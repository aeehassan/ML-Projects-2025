# Data preprocessing

### Duplicates

```python
df.duplicated().sum()
```

```python
df.drop_duplicates(inplace=True)
```

### Text preprocessing

```python
import re
import contractions
import string
from nltk.tokenize import RegexpTokenizer
```

```python
def prepped_text(text):
    # To make sure input is a string
    text = str(text)
    # Turn to lowercase
    text = text.lower()
    # Removing links if any
    text = re.sub(r'http\S+|www\S+|https\S+', '', text) # sub - substitutes
    # Split numbers in text
    text = re.sub(r'(\d)([A-Za-z])', r'\1 \2', text)   # number before letter
    text = re.sub(r'([A-Za-z])(\d)', r'\1 \2', text) # number after letter
    # Replace linebreaks/tabs with spaces 
    text = re.sub(r'\s+', ' ', text)
    # Replace contractions w/full words
    text = contractions.fix(text)
    # Removing punctuations 
    punctuations = string.punctuation
    for i in punctuations:
        text = text.replace(i, '')
    ## Tokenize text into a variable -- excluding only numbers 
    tokeniser = RegexpTokenizer(r'\b[a-zA-Z]{2,}\b')
    return text
```

```python
df['cleaned_text'] = df['clean_comment'].apply(prepped_text)
```

```python
df.drop('clean_comment', axis=1, inplace=True)
```

```python
df
```

### Vectorizer

```python
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
vectorizer = TfidfVectorizer(stop_words='english', max_features=5000) #
```

```python
X = vectorizer.fit_transform(df['cleaned_text'])
```

# Data splitting

I checked for class imbalance before splitting. After I confirmed it, I applied SMOTE technique to oversample the data to a more reasonable percentage distribution.

```python
from sklearn.model_selection import train_test_split
```

### Class imbalance 

```python
y = df['category']
```

```python
y.value_counts()
```

```python
y.value_counts(normalize=True)
```

```python
from imblearn.over_sampling import SMOTE
```

```python
sm = SMOTE(sampling_strategy={0: 15500, -1: 15500}, random_state=300)
X_res, y_res = sm.fit_resample(X, y)
```

```python
y_res.value_counts()
```

```python
y_res.value_counts(normalize=True)
```

```python
X_train, X_test, y_train, y_test = train_test_split(X_res, y_res, test_size=0.2, random_state=300, stratify=y_res)
```

# Model building

```python
from sklearn.svm import SVC
```

```python
%%time
model = SVC(kernel='linear') #
model.fit(X_train, y_train)
```

### Hyperparameter tuning 

```python
from sklearn.model_selection import GridSearchCV
```

```python
# Define parameter grid
param_grid = {
    'C': [0.01, 0.1, 1.0, 1.2, 1.5, 2, 2.5, 3, 3.5, 4, 5, 10, 100],
    'kernel': ['linear']
    # Data is linearly seperable so you don't need rbf
}
```

```python
# GridSearchCV
grid_search = GridSearchCV(
    estimator=model,
    param_grid=param_grid,
    scoring='accuracy',   # optimize for accuracy
    cv=5,                 # 5-fold cross-validation
    verbose=3,
    n_jobs=-1             # use all available CPU cores
)
```

```python
grid_search.fit(X_train, y_train)
```

```python
print("Best Parameters:", grid_search.best_params_)
```

# Evaluation

```python
from sklearn.metrics import classification_report
```

```python
y_pred = grid_search.predict(X_test)
```

```python
print(classification_report(y_test, y_pred))
```

```python
print(classification_report(y_test, model.predict(X_test)))
```
