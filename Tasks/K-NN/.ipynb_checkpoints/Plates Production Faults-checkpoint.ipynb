### Outlier removal

The box plots reveal **three categories of outliers** based on severity:

- **Extreme Outliers** – replaced with values at the **1st percentile**  
- **Medium Outliers** – replaced with values at the **2.5th percentile**  
- **Mild Outliers** – replaced with values at the **5th percentile**  

> **Reasoning:**  
> The farther an outlier is from the main data distribution,  
> the closer its replacement should be to the lower percentiles.  
> Akso, to improve model performance, these different categories
> must be filled differently

```python
# Outliers
extreme = ['v3','v4','v5','v6','v7','v8','v18']
medium = ['v10','v14','v26']
mild = ['v9','v16','v22','v23','v24']

# All cols
cols = df.columns.drop('target')
```

```python
# Replace them based on extremity of values
from scipy.stats.mstats import winsorize

for col in cols:
    if col in extreme:
        df[col] = winsorize(df[col], limits=[0.01, 0.01])
    elif col in medium:
        df[col] = winsorize(df[col], limits=[0.025, 0.025])
    else: 
        df[col] = winsorize(df[col], limits=[0.05, 0.05])
```

# Model building

### Model Training and Initial Evaluation

The model was first trained using all available features, with an arbitrary value of **k** chosen to evaluate baseline performance.  

The initial evaluation resulted in an accuracy of **70%**, indicating a need for improvement. To enhance performance, several techniques were attempted, including **feature selection**, **feature extraction**, and **dimensionality reduction**.

---

### Feature Selection & Dimensionality Reduction

- **Recursive Feature Elimination (RFE):**  
  Ineffective because K-NN does not provide feature coefficients required for ranking.

- **SelectKBest:**  
  Resulted in poorer performance than the baseline (accuracy dropped below 70%).

- **VarianceThreshold:**  
  Showed no effect since all 27 features were retained.

- **Principal Component Analysis (PCA):**  
  Provided the best results, reducing the feature space to **20 components** and improving metrics to the **mid-to-high 70s**, peaking at **79%**.

---

### Addressing Class Imbalance with SMOTE

To handle class imbalance, the **SMOTE** technique was applied. Various sampling values were tested for minority classes until optimal counts were achieved:

- Class **1**: **250 samples**  
- Class **5**: **200 samples**

This adjustment stabilized model accuracy at **74%**, a modest improvement over the baseline but still leaving room for further improvement.

```python
from sklearn.neighbors import KNeighborsClassifier
```

```python
model = KNeighborsClassifier(n_neighbors=3)
model.fit(X_train_r, y_train_r)
```

```python
y_train_r.value_counts()
```

### Refined KNN

This stage involves retraining the model using the **PCA-transformed dataset**, but with a focus on picking the optimal value of **k**. To achieve this, the **Elbow Method** is applied.  

The Elbow Method works by plotting a graph of **k values** against **accuracy**, allowing us to evaluate model performance across different **k** settings. The point where the accuracy begins to fall (the "elbow") represents the optimal **k** value.  

Once identified, this optimal value of **k** is then passed as a parameter to the model before final training and fitting.

#### Elbow method 

```python
k_values = range(1,21)
accuracies = []

for k in k_values:
    model.n_neighbors = k
    model.fit(X_train_r, y_train_r)
    accuracies.append(model.score(X_test, y_test))

plt.plot(k_values, accuracies, marker='o')
plt.xlabel('K neighbors')
plt.ylabel('Accuracy')
plt.title('Elbow method for Optimal K')
plt.grid(True)
plt.show()
```

```python
model.n_neighbors = 5
model.fit(X_train_r, y_train_r)
```

# Model evaluation

### Model Evaluation for Multi-Class Classification

The model is then evaluated. Unlike **binary classification models**, this is a **multi-class classification** task.  
Therefore, we must specify the `average` parameter to determine how the metrics should be combined across all classes.  
This provides a more representative measure of overall model performance.

Finally, the **confusion matrix** is transposed so that:
- **Actual values** appear on the **x-axis**  
- **Predicted values** appear on the **y-axis**  

This orientation improves interpretability and makes the results easier to analyze.

```python
from sklearn.metrics import precision_score, classification_report, recall_score, f1_score, confusion_matrix, ConfusionMatrixDisplay
```

```python
y_pred = model.predict(X_test)
```

```python
acc = model.score(X_test, y_test)
pr = precision_score(y_test, y_pred, average='macro')
re = recall_score(y_test, y_pred, average='macro')
f1 = f1_score(y_test, y_pred, average='macro')
cr = classification_report(y_test, y_pred)
```

```python
print(f'Accuracy: {acc:.3f}')
print(f'Precision: {pr:.2f}')
print(f'Recall: {re:.2f}')
print(f'F1 score: {f1:.2f}')
print(cr)
```

```python
cmx = confusion_matrix(y_test, y_pred)
cmx = cmx.T
disp = ConfusionMatrixDisplay(confusion_matrix=cmx)
disp.plot(cmap='Greens', values_format='d')
plt.xlabel('Actual values')
plt.ylabel('Predicted values')
plt.title('Confusion matrix', fontsize=14)
```
