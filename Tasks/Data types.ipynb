### Adding the new keys: Email, Age, and CGPA

```python
# Add Email key for each student using for loop
# The loop names the email based on the student's position

position = 1
for student in srs:
    srs[student]['Email'] = f'student{position}@gmail.com'
    position += 1

srs
```

```python
# Add Age key for each student

# Create a list to store each age
Ages = [20,20,22,24,21,
        20,21,20,20,20,
        22,22,23,20,18,
        20,20,23,21,24]
# Parse the ages as values into their respective keys
index = 0
for student in srs:
    srs[student]['Age'] = Ages[index]
    index += 1

srs
```

```python
# Add CGPA key for each student

CGPAs = [4.99,4.97,4.50,4.48,4.49,
               4.85,3.89,4.98,4.88,4.32,
               4.42,4.05,4.23,4.33,4.50,
               4.99,4.97,3.95,4.89,4.85]

# Parse the CGPAs as values into their respective keys
index = 0
for student in srs:
    srs[student]['CGPA'] = CGPAs[index]
    index += 1

srs
```
