# Feature engineering

```python
df.info()
```

```python
df['start_time'] = pd.to_datetime(df['start_time'])
```

```python
df['day'] = df['start_time'].dt.dayofweek
```

```python
df['hour'] = df['start_time'].dt.hour
```

```python
## 5 & 6 represent Saturday and Sunday respectively
df['day_named'] = df['day'].apply(lambda x: 'Weekend' if x > 4 else 'Weekday')
```
